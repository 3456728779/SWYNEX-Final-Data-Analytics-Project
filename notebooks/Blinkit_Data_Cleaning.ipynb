## Blinkit Data Cleaning and Preparation

 Step 1: Import libraries

```python
import pandas as pd
import numpy as np
```

Step 2: Load the dataset

```python
df = pd.read_csv("../data/blinkit_dataset.csv")

df.head()
```

Step 3: Understand the dataset

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

Step 4: Check missing values

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df["offer_type"] = df["offer_type"].fillna("No Offer")
```

```python
df["offer_type"].isnull().sum()
```

Step 5: Check duplicate records

```python
duplicate_count = df.duplicated().sum()
print("Duplicate Records:", duplicate_count)
```

Step 6: Check incorrect data types

```python
df.dtypes
```

```python
df["date_added"] = pd.to_datetime(df["date_added"])
df["expiry_date"] = pd.to_datetime(df["expiry_date"])
```

```python
df.dtypes
```

Step 7: Check inconsistent categorical values

```python
print(df["category"].unique())
print(df["brand"].unique())
print(df["city"].unique())
print(df["seller"].unique())
print(df["packaging_type"].unique())
print(df["offer_type"].unique())
print(df["delivery_status"].unique())
```

```python
df["category"].value_counts()
```

Step 8: Remove unnecessary spaces

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

Step 9: Check numerical values

```python
df["rating"].min(), df["rating"].max()
```

```python
df["discount_pct"].min(), df["discount_pct"].max()
```

```python
df["price"].min(), df["price"].max()
```

Step 10: Check logical consistency

```python
calculated_price = (
    df["price"] * (1 - df["discount_pct"] / 100)
).round(2)
```

```python
df["price_check"] = calculated_price == df["final_price"]

df["price_check"].value_counts()
```

Step 11: Check dates

```python
invalid_dates = (df["expiry_date"] < df["date_added"]).sum()
print("Invalid expiry dates:", invalid_dates)
```

Step 12: Remove the temporary checking column

```python
df.drop(columns=["price_check"], inplace=True)
```

Step 13: Final validation

```python
print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

Step 14: Save cleaned dataset

```python
df.to_csv("../data/Blinkit_Cleaned_Data.csv", index=False)

print("Cleaned dataset saved successfully.")
```
